```python
class BTNode:
  def __init__(self, elem):
    self.elem = elem
    self.right = None
    self.left = None
```

```python
def inorder(root):
  if root == None:
    return

  inorder(root.left)
  print(root.elem, end = ' ')
  inorder(root.right)
```

```python
def tree_construction(arr, i = 1):
  if i>=len(arr) or arr[i] == None:
    return None
  p = BTNode(arr[i])
  p.left = tree_construction(arr, 2*i)
  p.right = tree_construction(arr, 2*i+1)
  return p


root2 = tree_construction([None, 'A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', None, None, None, 'I', 'J', None, 'k'])
inorder(root2)
```

```python
def find_Path(root, key):
  p=[]
  p.append(root.elem)
  while root.elem!=key:
      if key>root.elem and root.right!=None:
        root=root.right
        p.append(root.elem)
      elif key<root.elem and root.left!=None:
        root=root.left
        p.append(root.elem)
      else:
        return "No Path Found"
  return p

#DRIVER CODE
root=BTNode(30)
a1=BTNode(10)
a2=BTNode(3)
a3=BTNode(15)
a5=BTNode(40)
a6=BTNode(35)
a7=BTNode(55)
root.left=a1
a1.left=a2
a1.right=a3
root.right=a5
a5.left=a6
a5.right=a7
#Write by yourself from the given tree
print(find_Path(root,15))
#This should print [30,10,15]

print(find_Path(root,50))
#This should print No Path Found
```
