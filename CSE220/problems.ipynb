```python
class Node:
  def __init__(self, val, left, right):
    self.val = val
    self.left = left
    self.right = right
def createTree(arr, i):
  root = None
  if i>0 and i < len(arr):
    if arr[i] != None:
      root = Node(arr[i], None, None)
      root.left = createTree(arr, 2*i)
      root.right = createTree(arr, 2 * i + 1)
  return root
x=[None, 19, 22, 10, 32, 20, None, 30, None, 5, None, 7, None, None, 2, -2]
y=createTree(x,1)

def postOrder(root):
  if root != None:
    postOrder(root.left)
    postOrder(root.right)
    print(root.val, end=" ")

def preOrder(root):
  if root != None:
    print(root.val, end=" ")
    preOrder(root.left)
    preOrder(root.right)
print("preorder",end=" ")
preOrder(y)
print("\n")
print("my preorder: 19 22 32 5 2 -2 20 7 10 30")
print("\n")
print("\n")
print("postorder",end=" ")
postOrder(y)
print("\n")
print("my postorder: 2 -2 5 32 7 20 22 30 10 19")
```
