```python
#Task 01: Merge Lineup
import numpy as np
def mergeLineup(pokemon_1, pokemon_2):
  res=np.array([0]*len(pokemon_1))
  for i in range(len(pokemon_1)):
    if pokemon_1[i]!=None and pokemon_2[len(pokemon_1)-i-1]!=None:
      res[i]=pokemon_1[i]+pokemon_2[len(pokemon_1)-i-1]
    elif pokemon_1[i]==None:
      res[i]=0+pokemon_2[len(pokemon_1)-i-1]
    elif pokemon_2[len(pokemon_1)-i-1]==None:
      res[i]=pokemon_1[i]+0
    else:
      res[i]=0
  return res
print("///  Task 01: Merge Lineup  ///")
pokemon_1 = np.array([12, 3, 25, 1, None])
pokemon_2 = np.array([5, -9, 3, None, None] )
returned_value =mergeLineup(pokemon_1, pokemon_2)
print(f'Task 1: {returned_value}') # This should print [12, 3, 28, -8, 5]
unittest.output_test(returned_value, np.array([12, 3, 28, -8, 5]))
pokemon_1 = np.array([4, 5, -1, None, None])
pokemon_2 = np.array([2, 27, 7, 12, None])
returned_value =mergeLineup(pokemon_1, pokemon_2)
print(f'Task 1: {returned_value}') # This should print [4,17,6,27,2]
unittest.output_test(returned_value, np.array([4,17,6,27,2]))
```

```python
# Task 02: Discard Cards
def discardCards(cards, t):
  res=np.array([0]*len(cards))
  o=0
  k=0
  for i in range(len(cards)):
    if cards[i]==t:
      if k==0:
        k=1
      else:
        res[o]=cards[i]
        o+=1
        k=0
    else:
      res[o]=cards[i]
      o+=1
  return res
print("///  Task 02: Discard Cards  ///")
cards = np.array([1,3,7,2,5,2,2,2,0])
returned_value = discardCards(cards, 2)
print(f'Task 2: {returned_value}') # This should print [1,3,7,5,2,2,0,0,0]
unittest.output_test(returned_value, np.array([1,3,7,5,2,2,0,0,0]))

cards = np.array([5,5,5,0,0])
returned_value = discardCards(cards, 5)
print(f'Task 2: {returned_value}') # This should print [5,0,0,0,0]
unittest.output_test(returned_value, np.array([5,0,0,0,0]))
```

```python
#Task 03: Decryption Process
import numpy as np
def decrypt_matrix(matrix):
  res=np.array([0]*(len(matrix[0])-1))
  sum=np.array([0]*(len(matrix[0])))
  for c in range(len(matrix[0])):
    for r in range(len(matrix)):
      sum[c]+=matrix[r][c]
  j=1
  for i in range(len(res)):
    res[i]=sum[j]-sum[i]
    j+=1
  return res
matrix=np.array([[1,3,1],
                 [6,4,2],
                 [5,1,7],
                 [9,3,3],
                 [8,5,4]
                 ])

returned_array=decrypt_matrix(matrix)
print(returned_array)
#This should print [-13, 1]
```

```python
#Task 06: Matrix Compression
def compress_matrix(mat):
  r=np.zeros([len(mat)//2,len(mat[0])//2],dtype="int")
  i=0
  k=0
  for j in range(len(r)):
    k=0
    for l in range(len(r[0])):
      r[j][l]+=mat[i][k]+mat[i][k+1]+mat[i+1][k]+mat[i+1][k+1]
      k+=2
    i+=2
  return r
b=[[1,2,3,4],[5,6,7,8],[1,3,5,2],[-2,0,6,-3]]
compress_matrix(b)

matrix=np.array([[1,2,3,4],
                 [5,6,7,8],
                 [1,3,5,2],
                 [-2,0,6,-3]
                 ])
print_matrix(matrix)
print("...............")
returned_array=compress_matrix(matrix)
print_matrix(returned_array)
#This should print

#|  14  |  22 |
#--------------
#|  2  |  10  |
#--------------
```
