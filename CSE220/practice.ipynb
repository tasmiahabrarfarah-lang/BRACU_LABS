```python
class Node:
  def __init__(self, val, left, right):
    self.val = val
    self.left = left
    self.right = right
def createTree(arr, i):
  root = None
  if i>0 and i < len(arr):
    if arr[i] != None:
      root = Node(arr[i], None, None)
      root.left = createTree(arr, 2*i)
      root.right = createTree(arr, 2 * i + 1)
  return root
x=[None, 19, 22, 10, 32, 20, None, 30, None, 5, None, 7, None, None, 2, -2]
y=createTree(x,1)

def postOrder(root):
  if root != None:
    postOrder(root.left)
    postOrder(root.right)
    print(root.val, end=" ")

def preOrder(root):
  if root != None:
    print(root.val, end=" ")
    postOrder(root.left)
    postOrder(root.right)
preOrder(y)
print("\n")
postOrder(y)
```

```python
import numpy as np
def column(m,c):
  s=np.zeros(len(m),dtype=int)
  for i in range(len(m)):
    s[i]=m[i][c]
  return s
def rotate(m):
  s=len(m)
  c0=column(m,0)
  cl=column(m,s-1)
  r0=np.array(m[0])
  rl=np.array(m[s-1])
  for i in range(s):
    m[s-1-i][0]=r0[i]
  for i in range(s):
    m[i][s-1]=rl[s-1-i]
  for i in range(1,s-1):
    m[s-1][i]=c0[i]
  for i in range(1,s-1):
    m[0][i]=cl[i]
  return m
mtrx=np.array([[1,2,3],[5,6,7],[9,10,11]])
print(mtrx)
print(rotate(mtrx))
```
