```python
#4.matrix
import numpy as np
class graph:
  def __init__(self,a):
    self.arr=a
    self.x=len(a)
    for i in range(self.x):
      b=np.zeros((self.x,self.x),dtype=int)
    self.m=b
  def ac(self,a,b,w=1):
    self.m[a][b]=w
  def undirected(self,a):
    n=len(a[0])
    newa=np.zeros([n,n],dtype=int)
    for r in range(n):
      for c in range(n):
        if a[r][c]!=0:
          newa[r][c]=a[r][c]
          newa[c][r]=a[r][c]
    return newa
a=[0,1,2,3,4,5,6]
x=graph(a)
x.ac(0,0,10)
x.ac(0,1,20)
x.ac(0,2,50)
x.ac(3,0,40)
x.ac(4,2,70)
x.ac(4,1,70)
x.ac(3,2,30)
x.ac(4,5,50)
x.ac(5,6,60)
x.ac(6,6,20)
x.ac(3,4,100)
x.ac(1,5,30)
x.ac(2,2,10)
x.ac(2,6,80)
x.ac(2,1,60)
x.ac(6,4,40)
print(x.m)
b=x.m
print(x.undirected(b))
```

```python
#4.list
import numpy as np
class Node:
  def __init__(self,a,w=1):
    self.val=a
    self.w=w
    self.next=None
class graph:
  def __init__(self,a):
    self.arr=a
    self.x=len(a)
    for i in range(self.x):
      self.arr[i]=Node(i)
  def undirected(self, a):
    newa = [None] * len(a)
    for i in range(len(a)):
      newa[i] = Node(a[i].val)
    for i in range(len(a)):
      temp = a[i].next
      while temp!=None:
        c1 = newa[i]
        while c1.next!=None:
          c1 = c1.next
        if temp.val!=newa[i].val:
          c1.next = Node(temp.val, temp.w)
        c2 = newa[temp.val]
        while c2.next!=None:
          c2 = c2.next
        c2.next = Node(i, temp.w)
        temp = temp.next
    return newa
def printall(a):
  for i in range(len(a)):
    temp=a[i]
    while temp!=None:
      print(f"{temp.val}({temp.w})",end="=>")
      temp=temp.next
    if temp==None:
      print("")
a=[0,1,2,3,4,5,6]
x=graph(a)

x.arr[0].next=Node(0,10)
x.arr[0].next.next=Node(1,20)
x.arr[0].next.next.next=Node(2,50)

x.arr[1].next=Node(5,30)

x.arr[2].next=Node(1,60)
x.arr[2].next.next=Node(6,80)
x.arr[2].next.next.next=Node(2,10)

x.arr[3].next=Node(0,40)
x.arr[3].next.next=Node(2,30)
x.arr[3].next.next.next=Node(4,100)

x.arr[4].next=Node(2,70)
x.arr[4].next.next=Node(5,50)
x.arr[4].next.next.next=Node(1,70)

x.arr[5].next=Node(6,60)

x.arr[6].next=Node(6,20)
x.arr[6].next.next=Node(4,40)

printall(x.arr)
print("\n")
y=x.undirected(x.arr)
printall(y)
```
