**Assignment Part 1: Doubly Linked List**

For Assignment Part 1, you should write your full code in the following cells along with your driver codes on your own.

```python
#Assignment Part 1

class Patient:
  #write a constructor
  def __init__(self, id, name, age, bloodgroup, next, prev):
    self.id = id
    self.name = name
    self.age = age
    self.bloodgroup = bloodgroup
    self.next = next
    self.prev = prev
```

```python
class WRM:

  def __init__(self):
    #Creating the dummy head
    self.dh = Patient(None,None,None,None,None,None)
    self.dh.next = self.dh
    self.dh.prev = self.dh

  def registerPatient(self,id,name,age,bloodgroup):
    po=Patient(id,name,age,bloodgroup,None,None)
    tail=self.dh.prev
    tail.next=po
    po.prev=tail
    po.next=self.dh
    self.dh.prev=po
    print("Registration successful!")

  def servePatient(self):
    tail=self.dh.next
    if self.dh.next!=None and self.dh.prev!=None:
      print(f"Served {tail.name}")
      self.dh.next=tail.next
      tail.next.prev=self.dh
      tail.next=None
      tail.prev=None
    else:
      print("No patients waiting.")

  def showAllPatient(self):
    tail=self.dh.next
    if tail!=self.dh:
      while tail.name!=self.dh.name:
        print(tail.name)
        tail=tail.next
    else:
      print("Line empty")

  def canDoctorGoHome(self):
    if self.dh.next==self.dh and self.dh.prev==self.dh:
      print("True")
    else:
      print("False")

  def cancelAll(self):
    self.dh.next=self.dh
    self.dh.prev=self.dh
    print("All appointments cancelled!")

  def ReverseTheLine(self):
    temp = self.dh.next
    while temp != self.dh:
        temp.next, temp.prev = temp.prev, temp.next
        temp = temp.prev
    self.dh.next, self.dh.prev = self.dh.prev, self.dh.next
    print("Successfully reversed!")
```

```python
#Write a Tester Code in this cell
print("**Welcome to Waiting Room Management System**")
x=WRM()
start=0
while start!=7:
  print("Enter 1 to register a patient\nEnter 2 to serve the first patient\nEnter 3 to show all patients\nEnter 4 to check whether doctor can go home\nEnter 5 to cancel all appointments\nEnter 6 to reverse the line\nEnter 7 to quit.")
  start=int(input())
  if start==1:
    id=input("Enter id:")
    name=input("Enter name:")
    age=input("Enter age:")
    bg=input("Enter blood group:")
    x.registerPatient(id,name,age,bg)
  elif start==2:
    x.servePatient()
  elif start==3:
    x.showAllPatient()
  elif start==4:
    x.canDoctorGoHome()
  elif start==5:
    x.cancelAll()
  elif start==6:
    x.ReverseTheLine()
  else:
    print("Error")
```

**Assignment Part 2: Stack**

Linked List based Stack is implemented in the following cell.

```python
class Node:
  def __init__(self,elem=None,next=None):
    self.elem = elem
    self.next = next

class Stack:
  def __init__(self):
    self.__top = None

  def push(self,elem):
    nn = Node(elem,self.__top)
    self.__top = nn

  def pop(self):
    if self.__top == None:
      #print('Stack Underflow')
      return None
    e = self.__top
    self.__top = self.__top.next
    return e.elem

  def peek(self):
    if self.__top == None:
      #print('Stack Underflow')
      return None
    return self.__top.elem

  def isEmpty(self):
    return self.__top == None
```

Task 1: Diamond Count

```python
def diamond_count(stack,string):
  c=0
  for i in string:
    if i=="<":
      stack.push(i)
    elif i==">":
      x=stack.pop()
      if x=="<":
        c+=1
  return c
print('Test 01')
stack = Stack()
string = '<..><.<..>> '
returned_value = diamond_count(stack,string)
print(f'Number of Diamonds: {returned_value}') #This should print 3
unittest.output_test(returned_value, 3)
print('-----------------------------------------')


print('Test 02')
stack = Stack()
string = '<<<..<......<<<<....>'
returned_value = diamond_count(stack,string)
print(f'Number of Diamonds: {returned_value}') #This should print 1
unittest.output_test(returned_value, 1)
print('-----------------------------------------')


print('Test 03')
stack = Stack()
string = '>>><...<<..>>...>...>>>'
returned_value = diamond_count(stack,string)
print(f'Number of Diamonds: {returned_value}') #This should print 3
unittest.output_test(returned_value, 3)
print('-----------------------------------------')
```

Task 2: Tower of Blocks

```python
def remove_block(stack, n):
  y=Stack()
  for i in range(n):
    x=stack.pop()
    y.push(x)
  y.pop()
  for i in range(1,n):
    z=y.pop()
    stack.push(z)


print('Test 01')
st = Stack()
st.push(4)
st.push(19)
st.push(23)
st.push(17)
st.push(5)
print('Stack:')
print_stack(st)
print('------')
remove_block(st,2)
print('After Removal')
print_stack(st)
print('------')

print()
print('======================================')
print()

print('Test 02')
st = Stack()
st.push(73)
st.push(85)
st.push(15)
st.push(41)
print('Stack:')
print_stack(st)
print('------')
remove_block(st,3)
print('After Removal')
print_stack(st)
print('------')

print()
print('======================================')
print()
```

Task 3: Stack Reverse

```python
def conditional_reverse(stack):
  x=Stack()
  while stack.peek()!=None:
    if x.peek()!=stack.peek():
      y=stack.pop()
      x.push(y)
      while x.peek()==stack.peek():
        stack.pop()
  return x


print('Test 01')
st=Stack()
st.push(10)
st.push(10)
st.push(20)
st.push(20)
st.push(30)
st.push(10)
st.push(50)
print('Stack:')
print_stack(st)
print('------')
reversed_stack=conditional_reverse(st)
print('Conditional Reversed Stack:')
print_stack(reversed_stack) # This stack contains 50, 10, 30, 20, 10 in this order whereas top element should be 10
print('------')
```
