```python
#1
d1=input()
d2=input()
d1n=d1.split(', ')
d2n=d2.split(', ')
k1=[]
k2=[]
v1=[]
v2=[]
ldk=[]
ldv=[]
ld={}
vt=()
nv=[]
for i in d1n:
  k1.append(i[0])
  v1.append(int(i[3::]))
for i in d2n:
  k2.append(i[0])
  v2.append(int(i[3::]))
for i in range(len(k1)):
  for j in range(len(k1)):
    if k1[i] in k2[j]:
      ldk.append(k1[i])
      v1[i]+=v2[j]
      ldv.append(v1[i])
    else:
      ldk.append(k1[i])
      ldv.append(v1[i])
for i in range(len(k2)):
  if k2[i] not in ldk:
    ldk.append(k2[i])
    ldv.append(v2[i])
for i in range(len(ldk)):
  ld[ldk[i]]=ldv[i]
for i in ld:
  if ld[i] not in nv:
    nv.append(ld[i])
vt=tuple(sorted(nv))
print(ld)
print('Values:',vt)
```

```python
#2
d={}
nd={}
while True:
  inp=input()
  if inp=='STOP':
    break
  else:
    d[inp]=0
  if inp not in nd:
    nd[inp]=1
  else:
    nd[inp]+=1
for i in nd:
  print(i,'-',nd[i],'times')
```

```python
#3
d={}
kv=[]
kvi=[]
inp=input()
kv=inp.split(', ')
kvi={}
for i in kv:
  a,b=i.split(" : ")
  kvi[a]=b
for i,j in kvi.items():
  if j not in d:
    d[j]=[i]
  else:
    d[j].append(i)
print(d)
```
