```python
#10
class Student:
    def __init__(self,name,ID):
        self.name = name
        self.ID = ID
    def Details(self):
        return "Name: "+self.name+"\n"+"ID: "+self.ID+"\n"
#Write your code here
class CSEStudent(Student):
  def __init__(self,n,i,s):
    super().__init__(n,i)
    self.s=s
    self.cl=[]
    self.ml=[]
    self.cd={}
  def Details(self):
    return super().Details()+"Current semester:"+self.s
  def addCourseWithMarks(self,*args):
    for i in range(len(args)):
      if i%2==0:
        self.cl.append(args[i])
      else:
        if args[i]>=85:
          self.ml.append(4.0)
        elif 80<=args[i]<=84:
          self.ml.append(3.3)
        elif 70<=args[i]<=79:
          self.ml.append(3.0)
        elif 65<=args[i]<=69:
          self.ml.append(2.3)
        elif 57<=args[i]<=64:
          self.ml.append(2.0)
        elif 55<=args[i]<=56:
          self.ml.append(1.3)
        elif 50<=args[i]<=54:
          self.ml.append(1.0)
        elif args[i]<50:
          self.ml.append(0.0)
    for i in range(len(self.cl)):
      self.cd[self.cl[i]]=self.ml[i]
  def showGPA(self):
    sum=0
    print(f"{self.name} has taken {len(self.cl)} courses.")
    for i in self.cd:
      print(f"{i}: {self.cd[i]}")
      sum+=self.cd[i]*3
    gpa=round(sum/(len(self.cd)*3),2)
    print(f"GPA of {self.name} is: {gpa}")

Bob = CSEStudent("Bob","20301018",'Fall 2020”')
Carol = CSEStudent("Carol","16301814",'”Fall 2020”')
Anny = CSEStudent("Anny","18201234",'”Fall 2020”')
print("#########################")
print(Bob.Details())
print("#########################")
print(Carol.Details())
print("#########################")
print(Anny.Details())
print("#########################")
Bob.addCourseWithMarks("CSE111",83.5,"CSE230",73.0,"CSE260",92.5)
Carol.addCourseWithMarks("CSE470",62.5,"CSE422",69.0,"CSE460",76.5,"CSE461",87.0)
Anny.addCourseWithMarks("CSE340",45.5,"CSE321",95.0,"CSE370",91.0)
print("----------------------------")
Bob.showGPA()
print("----------------------------")
Carol.showGPA()
print("----------------------------")
Anny.showGPA()
```
