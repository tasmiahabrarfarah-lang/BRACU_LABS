```python
#1
def BMI(height,weight):
  height=height/100
  bmi=round(weight/(height)**2,1)
  if bmi<18.5:
    return(f"Score is {bmi}. You are Underweight")
  elif 18.5<bmi<24.9:
    return(f"Score is {bmi}. You are Normal")
  elif 25<bmi<30:
    return(f"Score is {bmi}. You are Overweight")
  elif bmi>30:
    return(f"Score is {bmi}. You are Obese")
print(BMI(175,96))
print(BMI(152,48))
```

```python
#3
def replace_domain(email,new,old="kaaj.com"):
  n,d=email.split("@")
  if d!=new:
    d=new
    return(f"Changed:{n}@{d}")
  else:
    return(f"Unchanged:{n}@{d}")
print(replace_domain('alice@kaaj.com','sheba.xyz','kaaj.com'))
print(replace_domain('bob@sheba.com','sheba.xyz'))
```

```python
#4
def is_palindrome(inp):
  b=''
  new=''
  for i in range(-1,-len(inp)-1,-1):
   if inp[i]!=" ":
    b+=inp[i]
  for i in inp:
    if i!=" ":
      new+=i
  if b==new:
    return("Palindrome")
  else:
    return("Not a Palindrome")
print(is_palindrome("madam"))
print(is_palindrome("hello"))
print(is_palindrome("nurses run"))
```

```python
#5
def numbers(days):
  years=days//365
  ly=days-365*years
  months=ly//30
  lm=ly-30*months
  days=lm
  return(f"{years} years, {months} months and {days} days")
print(numbers(4320))
print(numbers(4000))
```

```python
#6
def func(inp):
  signs='.!?'
  space=' '
  new=''
  if ord("a")<ord(inp[0])<ord("z"):
    new+=chr(ord(inp[0])-32)
  for i in range(1,len(inp)):
    if inp[i-2] in signs and (i-2)>0:
      new+=chr(ord(inp[i])-32)
    elif inp[i-1]==" " and inp[i+1] in space and inp[i]=="i":
      new+="I"
    else:
      new+=inp[i]
  return new
print(func("my favourite animal is a dog. a dog has sharp teeth so that it can eat flesh very easily. do you know my pet dog's name? i love my pet very much."))
```

```python
#2
def is_james_bond(l):
  flag=False
  for i in range(len(l)):
    if l[i]==0:
      for j in range(i,len(l)):
        if l[j]==0:
          for k in range(j,len(l)):
            if l[k]==7:
              flag=True
  return flag
print(is_james_bond( [1, 2, 4, 0, 0, 7, 5] ))
print(is_james_bond( [1, 7, 2, 0, 4, 5, 0] ))
print(is_james_bond( [1, 0, 2, 0, 4, 7, 5] ))
```

```python
#3
def assign_students_to_sections(sections,*names):
  new={}
  for k in range(len(sections)):
    new[sections[k]]=[]
  for i in names:
    sum=0
    sec=0
    for j in i:
      sum+=ord(j)
    sec=sum%len(sections)
    for l in range(len(sections)):
      if sec==l:
        new[sections[l]].append(i)
  return new
assign_students_to_sections ('ABCDE', 'Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank', 'Grace')
```

```python
#4
def username_generator(first_name,last_name,student_id,middle_name=''):
  new=''
  if len(first_name)>=3:
    for i in range(3):
      if ord('a')<=ord(first_name[i])<=ord('z'):
        new+=chr(ord(first_name[i])-32)
      else:
        new+=first_name[i]
  else:
    for i in range(len(first_name)):
      if ord('a')<=ord(first_name[i])<=ord('z'):
        new+=chr(ord(first_name[i])-32)
      else:
        new+=first_name[i]
  if middle_name!='':
    new+=middle_name
  if len(last_name)>=3:
    for i in range(len(last_name)-3,len(last_name)):
      if ord('A')<=ord(first_name[i])<=ord('Z'):
        new+=chr(ord(last_name[i])+32)
      else:
        new+=last_name[i]
  else:
    for i in range(len(last_name)):
      if ord('A')<=ord(first_name[i])<=ord('Z'):
        new+=chr(ord(last_name[i])+32)
      else:
        new+=last_name[i]
  new+='_'
  student_id=str(student_id)
  for i in range(len(student_id)-4,len(student_id)):
    new+=student_id[i]
  return new
first_name, middle_name, last_name, student_id= input ("First Name:"), input("Middle Name:"), input ("Last Name:"), int (input ("Student ID:"))
print(username_generator(first_name, last_name, student_id, middle_name))
```

```python
#5
def key_generator(*names):
  keys=[]
  for i in names:
    key=''
    key+=chr(ord(i[0])+32)
    for j in range(-2,-len(i),-1):
      key+=str(ord(i[j]))
    key+=chr(ord(i[len(i)-1])-32)
    keys.append(key)
  return('Encrypted Keys:',keys)
key_list = key_generator ("Alex", "Bob", "Trudy")
print(key_list)
```

```python
#6
import random
def playRockPaperScissor(rounds):
  rounds=int(rounds)
  p=0
  c=0
  options=['rock','paper','scissor']
  for i in range(rounds):
    inp=input()
    comp=random.choice(options)
    print("Computer:",comp)
    if inp=='rock' and comp=='paper':
      c+=1
    if inp=='rock' and comp=='scissor':
      p+=1
    if inp=='paper' and comp=='scissor':
      c+=1
    if inp=='paper' and comp=='rock':
      p+=1
    if inp=='scissor' and comp=='paper':
      p+=1
    if inp=='scissor' and comp=='rock':
      c+=1
  print("Your Score:",p)
  print("Computer's Score:",c)
  if p>c:
    print("You have won the game!")
  elif p<c:
    print("Computer has won the game!")
  else:
    print("It's a draw!")
playRockPaperScissor(3)
```
