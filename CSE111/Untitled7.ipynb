```python
#1
class NikeBangladesh:
    cs = {"Air Jordan": 0, "Cortez": 0, "Zoom Kobe": 0}
    bo = []
    ts=0
    def __init__(self, l):
        self.p = l
        self.i = {"Air Jordan": 0, "Cortez": 0, "Zoom Kobe": 0}
        self.sold = 0
        NikeBangladesh.bo.append(l)
    def details(self):
        print(f"Nike {self.p} outlet:")
        print("Products Currently Stocked:")
        print(self.i)
        print(f"Sold: {self.sold}")
    @classmethod
    def status(cls):
        print("Nike Bangladesh Status:")
        print("Branches Opened:", cls.bo)
        print("Currently Stocked")
        print(cls.cs)
        print("Sold:", cls.ts)
    def restockProducts(self, d):
        for i,j in d.items():
            self.i[i] += j
            NikeBangladesh.cs[i] += j
    def productSold(self, d):
        for i,j in d.items():
                self.i[i] -= j
                self.sold += j
                NikeBangladesh.cs[i] -=j
                NikeBangladesh.ts+=j
print("xxxxxxxxxxxxxx1xxxxxxxxxxxxxxxx")
NikeBangladesh.status()
dhaka = NikeBangladesh("Dhaka Banani")
chittagong = NikeBangladesh("Chittagong GEC")
print("xxxxxxxxxxxxxx2xxxxxxxxxxxxxxxx")
dhaka.details()
print("xxxxxxxxxxxxxx3xxxxxxxxxxxxxxxx")
chittagong.details()
print("xxxxxxxxxxxxxx4xxxxxxxxxxxxxxxx")
dhaka.restockProducts({"Air Jordan": 1200, "Cortez": 200, "Zoom Kobe": 200})
chittagong.restockProducts({"Air Jordan": 1000, "Cortez": 250, "Zoom Kobe": 100})
print("xxxxxxxxxxxxxx5xxxxxxxxxxxxxxxx")
NikeBangladesh.status()
print("xxxxxxxxxxxxxx6xxxxxxxxxxxxxxxx")
dhaka.productSold({"Air Jordan": 760, "Cortez": 90})
chittagong.productSold({"Air Jordan": 520, "Zoom Kobe": 70})
print("xxxxxxxxxxxxxx7xxxxxxxxxxxxxxxx")
NikeBangladesh.status()
```

```python
#2
class Player:
  total=0
  pl=[]
  def __init__(self,name=None,jn=10,team=None):
    self.name=name
    self.team=team
    self.jn=jn
    Player.total+=1
    if self.name!=None:
      Player.pl.append(self.name)
  def set_name(self,t):
    self.name=t
    Player.pl.append(self.name)
  def set_team(self,t):
    self.team=t
  def set_number(self,t):
    self.jn=t
  def player_detail(self):
    return(f'Player Name: {self.name}\nJersey Number: {self.jn}\nCountry: {self.team}')
  @classmethod
  def info(cls):
    print('Total number of players:',cls.total)
    print('Players enlisted so far:',",".join(cls.pl))
print("Total number of players:", Player.total)
print("---------------------------")
p1 = Player()
p1.set_name("Neymar")
p1.set_team("Brazil")
print(p1.player_detail())
print('========================')
Player.info()
print("---------------------------")
p2 = Player("Ronaldo")
p2.set_number(7)
p2.set_team("Portugal")
print(p2.player_detail())
print('========================')
Player.info()
print("---------------------------")
p3 = Player("Messi")
p3.set_team("Argentina")
print(p3.player_detail())
print('========================')
Player.info()
print("---------------------------")
p4 = Player("Mbappe", 10, "France")
print(p4.player_detail())
print('========================')
Player.info()
```

```python
#3
class SultansDine:
  bn=0
  ts=0
  bd=[]
  @classmethod
  def details(cls):
    print('Total Number of branch(s):',cls.bn)
    print("Total Sell:",cls.ts,'Taka')
    for i in cls.bd:
      i.sp=round((i.sell/cls.ts)*100,2)
      print(f"Branch Name: {i.name}, Branch Sell: {i.sell}, Branch consists of total sell's: {i.sp}%")
  def __init__(self,name):
    self.name=name
    self.sell=0
    self.sp=0
    SultansDine.bd.append(self)
    SultansDine.bn+=1
  def sellQuantity(self,q):
    if q < 10:
      self.sell = q * 300
    elif q<20:
      self.sell = q * 350
    else:
      self.sell = q * 400
    SultansDine.ts+=self.sell
    self.sp=round((self.sell/SultansDine.ts)*100,2)
  def branchInformation(self):
    print("Branch Name:",self.name)
    print("Branch Sell:",self.sell,'Taka')
SultansDine.details()
print('########################')
dhanmondi = SultansDine('Dhanmondi')
dhanmondi.sellQuantity(25)
dhanmondi.branchInformation()
print('-----------------------------------------')
SultansDine.details()

print('========================')

baily_road = SultansDine('Baily Road')
baily_road.sellQuantity(15)
baily_road.branchInformation()
print('-----------------------------------------')
SultansDine.details()

print('========================')

gulshan = SultansDine('Gulshan')
gulshan.sellQuantity(9)
gulshan.branchInformation()
print('-----------------------------------------')
SultansDine.details()

```
