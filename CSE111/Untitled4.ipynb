```python
#2
class Farmer:
  def __init__(self,noid=None):
    self.noid=noid
    self.c=[]
    self.f=[]
    if self.noid==None:
      print("Welcome to your farm!")
    elif type(self.noid)==int:
      print(f"Welcome to your farm. Your farm ID is {self.noid}!")
    elif type(self.noid)==str:
      print(f"Welcome to your farm, {self.noid}!")
  def addCrops(self,*args):
    nc=0
    nc=len(args)
    if nc>0:
      print(f"{nc} crop(s) added.")
    else:
      print(f"No crop(s) added.")
    for i in args:
      self.c.append(i)
  def addFishes(self,*fargs):
    nf=0
    nf=len(fargs)
    if nf>0:
      print(f"{nf} fish(s) added.")
    else:
      print(f"No fish added.")
    for i in fargs:
      self.f.append(i)
  def showGoods(self):
    if len(self.c)>0:
      print(f"You have {len(self.c)} crop(s):")
    else:
      print(f"You don't have any crop(s).")
    for i in range(len(self.c)):
      if i!=len(self.c)-1:
        print(self.c[i],end=',')
      else:
        print(self.c[i])
    if len(self.f)>0:
      print(f"You have {len(self.f)} fish(s):")
    else:
      print(f"You don't have any fish(s).")
    for i in range(len(self.f)):
      if i!=len(self.f)-1:
        print(self.f[i],end=',')
      else:
        print(self.f[i])
f1 = Farmer()
print("-------------------")
f1.addCrops('Rice', "Jute", "Cinnamon")
print("-------------------")
f1.addFishes()
print("-------------------")
f1.addCrops('Mustard')
print("-------------------")
f1.showGoods()
print("-------------------")
f2 = Farmer("Korim Mia")
print("-------------------")
f2.addFishes('Pangash', 'Magur')
print("-------------------")
f2.addCrops("Wheat", "Potato")
print("-------------------")
f2.addFishes("Koi", "Tuna", "Sardine")
print("-------------------")
f2.showGoods()
print("-------------------")
f3 = Farmer(2865127000)
print("-------------------")
f3.addCrops()
print("-------------------")
f3.addFishes("Katla")
print("-------------------")
f3.showGoods()
print("-------------------")
```

```python
#3
class TaxiLagbe:
  def __init__(self,num,area):
    self.num=num
    self.area=area
    self.pl=[]
    self.nf=[]
    self.fare=0
    self.l=0
  def addPassenger(self,*args):
    nargs=[]
    for i in args:
      nargs.append(i.split("_"))
    self.l+=len(args)
    if self.l<=4:
      for i in args:
        self.nf.append(i.split("_"))
      for i in nargs:
        self.pl.append(i[0])
      for i in range(len(args)):
        print(f"Dear {nargs[i][0]}! Welcome to TaxiLagbe.")
      for i in range(len(nargs)):
        self.fare+=int(nargs[i][1])
    else:
      print("Taxi Full! No more passengers can be added.")
  def printDetails(self):
    print(f"Trip info for Taxi number: {self.num}")
    print(f"This taxi can only cover the {self.area} area.")
    print(f"Total passengers: {len(self.pl)}")
    print("Passenger lists:")
    for i in range(len(self.pl)):
      if i!=len(self.pl)-1:
        print(self.pl[i],end=', ')
      else:
        print(self.pl[i])
    print("Total collected fare:",self.fare)
taxi1 = TaxiLagbe('1010-01', 'Dhaka')
print('-------------------------------')
taxi1.addPassenger('Walker_100', 'Wood_200','Matt_100')
taxi1.addPassenger('Wilson_105')
print('-------------------------------')
taxi1.printDetails()
print('-------------------------------')
taxi1.addPassenger('Karen_200')
print('-------------------------------')
taxi1.printDetails()
print('-------------------------------')
taxi2 = TaxiLagbe('1010-02', 'Khulna')
taxi2.addPassenger('Ronald_115', 'Parker_215')
print('-------------------------------')
taxi2.printDetails()
```

```python
#4
class Sphere:
  def __init__(self,name,r=1,c="White"):
    self.name=name
    self.r=r
    self.c=c
    self.volume=4/3*3.1416*(self.r)**3
  def merge_sphere(self,*args):
    print("Spheres are being merged")
    for i in args:
      self.volume+=i.volume
    for i in args:
      if self.volume!=i.volume:
        self.c="Mixed Color"
  def printDetails(self):
    print("sphere ID:",self.name)
    print("Color:",self.c)
    print("Volume:",self.volume)
sphere1 = Sphere("Sphere 1")
print("1***************")
sphere1.printDetails()
print("2***************")
sphere2 = Sphere("Sphere 2", 3)
print("3***************")
sphere2.printDetails()
print("4***************")
sphere3 = Sphere("Sphere 3", 2)
print("5***************")
sphere3.printDetails()
print("6***************")
sphere3.merge_sphere(sphere1,sphere2)
print("7***************")
sphere3.printDetails()
print("8***************")
sphere4 = Sphere("Sphere 4", 5, "Purple")
print("9***************")
sphere4.merge_sphere(sphere3)
print("10***************")
sphere4.printDetails()
```
