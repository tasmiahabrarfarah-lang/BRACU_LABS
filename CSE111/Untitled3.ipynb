```python
#4
class StudentDatabase:
  def __init__(self,name,id):
    self.name=name
    self.id=id
    self.grades={}
    self.allc=[]
    self.semesters=[]
    self.g=[]
  def calculateGPA(self,grades1,sem):
    self.d={}
    self.grades2=[]
    self.courses=[]
    self.grade=0
    self.grades2=grades1
    self.sem=sem
    self.semesters.append(self.sem)
    for i in self.grades2:
      self.courses.append(i[0:6:1])
      self.grade+=float(i[7:len(i):1])*3
    self.grade=round(self.grade/(len(self.courses)*3),2)
    self.g.append(self.grade)
    self.allc.append(self.courses)
    self.courses=tuple(self.courses)
    self.d[self.courses]=self.grade
    self.grades[self.sem]=self.d
  def printDetails(self):
    print(f"Name: {self.name}\nID: {self.id}")
    for i in range(len(self.grades)):
      print(f"Courses taken in {self.semesters[i]}:")
      for k in range(len(self.allc[i])):
        print(self.allc[i][k])
      print("CGPA:",self.g[i])
s1 = StudentDatabase('Pietro', '10101222')
s1.calculateGPA(['CSE230: 4.0', 'CSE220: 4.0', 'MAT110: 4.0'], 'Summer2020')
s1.calculateGPA(['CSE250: 3.7', 'CSE330: 4.0'], 'Summer2021')
print(f'Grades for {s1.name}\n{s1.grades}')
print('---------------------------------')
s1.printDetails()
s2 = StudentDatabase('Wanda', '10103332')
s2.calculateGPA(['CSE111: 3.7', 'CSE260: 3.7', 'ENG101: 4.0'], 'Summer2022')
print('---------------------------------')
print(f'Grades for {s2.name}\n{s2.grades}')
print('---------------------------------')
s2.printDetails()

```

```python
#5 hw 1
class Student:
  def __init__(self,name,cg,c=9,dept="CSE"):
    self.name=name
    self.cg=cg
    self.c=c
    self.dept=dept
    self.status=''
  def checkScholarshipEligibility(self):
    if self.cg>=3.5 and self.c>10:
      if self.cg>=3.7:
        self.status='Merit-based scholarship'
        print(f"{self.name} is eligible for Merit-based scholarship.")
      elif self.cg<3.7:
        self.status='Need-based scholarship'
        print(f"{self.name} is eligible for Need-based scholarship.")
    else:
      self.status='No scholarship'
      print(f"{self.name} is not eligible for scholarship.")
  def showDetails(self):
    print("Name:",self.name)
    print("Department:",self.dept)
    print("CGPA:",self.cg)
    print("Number of Credits:",self.c)
    print("Scholarship Status:",self.status)
print('--------------------------')
std1 = Student("Alif", 3.99, 12)
print('--------------------------')
std1.checkScholarshipEligibility()
print('--------------------------')
std1.showDetails()
print('--------------------------')
std2 = Student("Mim", 3.4)
std3 = Student("Henry", 3.5, 15,"BBA")
print('--------------------------')
std2.checkScholarshipEligibility()
print('--------------------------')
std3.checkScholarshipEligibility()
print('--------------------------')
std2.showDetails()
print('--------------------------')
std3.showDetails()
print('--------------------------')
std4 = Student("Bob", 4.0, 6, "CSE")
print('--------------------------')
std4.checkScholarshipEligibility()
print('--------------------------')
std4.showDetails()


```

```python
#2
menu = {'Chicken Lollipop':15,'Beef Nugget':20,'Americano':180,'Red Velvet':150,'Prawn Tempura':80,'Saute Veg':200}
class Foodie:
  def __init__(self,name):
    self.name=name
    self.num=0
    self.items=[]
    self.ip=0
    self.bill=0
    self.inp=[]
    self.q=[]
    self.tips=0
  def show_orders(self):
    return(f"{self.name} has {self.num} item(s) in the cart\nItems:{self.items}\nTotal spent:{self.bill}.")
  def order(self,*args):
    np=[]
    items=[]
    q=[]
    for i in args:
      np.append(i.split("-"))
    for i in np:
      items.append(i[0])
      q.append(i[1])
    self.items.extend(items)
    for i in range(len(args)):
      if items[i] in menu.keys():
        self.ip=menu[items[i]]*int(q[i])
        print(f"Ordered- {items[i]}, quantity- {q[i]}, price (per unit)- {menu[items[i]]}.")
        print("Total price-",self.ip)
        self.bill+=self.ip
    print(self.bill)
    self.num=len(self.items)
  def pay_tips(self,tips=0):
    self.tips=tips
    self.bill+=self.tips
    if self.tips!=0:
      print(f"Gives {self.tips}/- tips to the waiter.")
    else:
      print("No tips to the waiter.")
f1 = Foodie('Frodo')
print(f1.show_orders())
print('1----------------------')
f1.order('Chicken Lollipop-3','Beef Nugget-6','Americano-1')
print('2----------------------')
print(f1.show_orders())
print('3----------------------')
f1.order('Red Velvet-1')
print('4----------------------')
f1.pay_tips(20)
print('5----------------------')
print(f1.show_orders())
f2 = Foodie('Bilbo')
print('6----------------------')
f2.order('Prawn Tempura-6','Saute Veg-1')
print('7----------------------')
f2.pay_tips()
print('8----------------------')
print(f2.show_orders())
```

```python
#3
class Department:
  def __init__(self,dept="ChE Department",sec=5):
    self.name=dept
    self.sec=sec
    self.avg=0
    self.lavg=[]
    self.lsec=[]
    self.ts=0
    self.favg=0
    print(f"The {self.name} Department has {self.sec} sections.")
  def add_students(self,*args):
    sum=0
    for i in args:
      sum+=i
    if len(args)==self.sec:
      self.avg=round(sum/self.sec,2)
      print(f"The {self.name} has an average of {self.avg} students in each section.")
    else:
      print(f"The {self.name} doesn't have {len(args)} sections.")
    self.ts=self.avg*self.sec
  def merge_Department(self,*args):
    for i in args:
      print(f"{i.name} is merged to {self.name}.")
      self.ts+=i.ts
    self.favg=round(self.ts/self.sec,1)
    return(f"Now the {self.name} has an average of {self.favg} students in each section.")
d1 = Department()
print('1-----------------------------')
d2 = Department('MME Department')
print('2-----------------------------')
d3 = Department('NCE Department', 8)
print('3-----------------------------')
d1.add_students(12, 23, 12, 34, 21)
print('4-----------------------------')
d2.add_students(40, 30, 21)
print('5-----------------------------')
d3.add_students(12, 34, 41, 17, 30, 22, 32, 51)
print('6-----------------------------')
mega = Department('Engineering Department', 10)
print('7-----------------------------')
mega.add_students(21,30,40,36,10,32,27,51,45,15)
print('8-----------------------------')
print(mega.merge_Department(d1, d2))
print('9-----------------------------')
print(mega.merge_Department(d3))
```

```python
#4
class Shopidify:
  def __init__(self,name='Guest'):
    self.name=name
    self.inpn=[]
    self.iq=[]
    if self.name!='Guest':
      print(f"Welcome {self.name} to Shopidify")
    else:
      print("Welcome to Shopidify")
  def add_to_cart(self,*args):
    print(type(args))
    for i in range(len(args)):
      if type(i)!=list:
        if len(args)%2==0:
          if i%2==0:
            if args[i] not in self.inpn:
              self.inpn.append(args[i])
            else:
              self.iq[i]+=1
          else:
            self.iq.append(args[i])
        else:
          if args[i] not in self.inpn:
            self.inpn.append(args[i])
            self.iq.append(1)
          else:
            for j in range(len(self.inpn)):
              if args[i]==self.inpn[j]:
                self.iq[j]+=1
      else:
          for j in range(len(i)):
            if len(i)%2==0:
              if j%2==0:
                if i[j] not in self.inpn:
                  self.inpn.append(i[j])
                else:
                  self.iq[j]+=1
              else:
                self.iq.append(i[j])
            else:
              if i[j] not in self.inpn:
                self.inpn.append(i[j])
                self.iq.append(1)
              else:
                for k in range(len(self.inpn)):
                  if i[j]==self.inpn[k]:
                    self.iq[k]+=1
  def display_cart(self):
    print(f"Items in the cart for {self.name}:")
    for i in range(len(self.inpn)):
      print(f'- {self.inpn[i]}: {self.iq[i]}x')
  def checkout(self):
    print(f"Checkout completed for {self.name}")
  def display_history(self):
    print(f"Purchase history for {self.name}:")
    print("Transaction 1:")
    for i in range(len(self.inpn)):
      print(f'- {self.inpn[i]}: {self.iq[i]}x')
guest_account = Shopidify()
print("1xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx")
john_account = Shopidify("John")
print("2xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx")
guest_account.add_to_cart("Air Jordan", 2)
guest_account.add_to_cart("Luffy Action Figure")
guest_account.display_cart()
print("3xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx")
john_account.add_to_cart(["Chocolate Chip Cookies", 3,"Goku Action Figure",2,"Dumbbells-5kg",2])
john_account.display_cart()
print("4xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx")
guest_account.add_to_cart("Air Jordan")
guest_account.display_cart()
print("5xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx")
guest_account.checkout()
print("6xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx")
guest_account.display_history()
print("7xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx")
john_account.checkout()
print("8xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx")
john_account.display_history()
print("9xxxxxxxxxxxxxxxxxxxxxxxxxxxxxxxx")
```
