```python
#2
class PokemonBasic:

  def __init__(self, name = 'Default', hp = 0, weakness = 'None', type = 'Unknown'):
    self.name = name
    self.hit_point = hp
    self.weakness = weakness
    self.type = type

  def get_type(self):
    return 'Main type: ' + self.type

  def get_move(self):
    return 'Basic move: ' + 'Quick Attack'

  def __str__(self):
    return "Name: " + self.name + ", HP: " + str(self.hit_point) + ", Weakness: " + self.weakness
class PokemonExtra(PokemonBasic):
  def __init__(self,n,hp,w,t,st=None,om=None):
    super().__init__(n,hp,w,t)
    self.st=st
    self.om=om
  def get_type(self):
    if self.st==None:
      return super().get_type()
    else:
      return(f'{super().get_type()}, Secondary type: {self.st}')
  def get_move(self):
    if self.om==None:
      return(super().get_move())
    else:
      ##Fixed the error
      return (f"{super().get_move()}\nOther move: {','.join(self.om)}")


print('\n------------Basic Info:--------------')
pk = PokemonBasic()
print(pk)
print(pk.get_type())
print(pk.get_move())

print('\n------------Pokemon 1 Info:-------------')
charmander = PokemonExtra('Charmander', 39, 'Water', 'Fire')
print(charmander)
print(charmander.get_type())
print(charmander.get_move())

print('\n------------Pokemon 2 Info:-------------')
charizard = PokemonExtra('Charizard', 78, 'Water', 'Fire', 'Flying', ('Fire Spin', 'Fire Blaze'))
print(charizard)
print(charizard.get_type())
print(charizard.get_move())
```

```python
#3
class Cakes:
  order_list={}
  reviews={}
  @classmethod
  def give_feedbacks(cls,name,r):
    print('Thanks for your valuable feedback!')
    if name not in cls.reviews.keys():
      cls.reviews[name]=[]
      cls.reviews[name].append(r)
    else:
      cls.reviews[name].append(r)
  @classmethod
  def show_feedbacks(cls):
    print(cls.reviews)
  def __init__(self,f,w):
    self.f=f
    self.w=w
    self.s='Moderate'
    self.c='Whipped Cream'
    self.p=self.w/1000*1200
    self.name=self.f+' Cake '+str(self.w)+'gm'
    if self.name not in Cakes.order_list.keys():
      Cakes.order_list[self.name]=1
    else:
      Cakes.order_list[self.name]+=1
  def cake_details(self):
    print(f"Flavor: {self.f} Cake, Weight: {self.w} gm")
    print(f"Sweetness: {self.s} sugar, Cream Type: {self.c}")
    print("Price:",self.p,'Taka')
  def add_customization(self,s,c):
    self.s=s
    self.c=c
class Cheese_Cakes(Cakes):
  def __init__(self,f,w,t='baked'):
    #Pass the f,w to parent's init
    self.f=f
    self.w=w
    self.name=self.f+' Cheese Cake '+str(self.w)+'gm'
    self.s='Moderate'
    self.t=t
    self.p=self.w/1000*1500
    self.c='Cream Cheese'
    #Reuse this code lines from the parent class
    if self.name not in Cakes.order_list.keys():
      Cakes.order_list[self.name]=1
    else:
      Cakes.order_list[self.name]+=1
  def cake_details(self):
    #Reuse the parent method
    print(f"Flavor: {self.f} Cake, Weight: {self.w} gm")
    print(f"Sweetness: {self.s} sugar, Cream Type: {self.c}")
    print(f"Cake Type:{self.t}, Price: {self.p} Taka")
  def add_customization(self):
    print("Sorry! No customization available for cheese cakes.")
order_1=Cakes("Chocolate",500)
order_2=Cakes("Vanilla",800)
print("(1)**********************************")
order_1.cake_details()
print("(1.1)**********************************")
print(Cakes.order_list)
print("(2)**********************************")
order_2.add_customization("Zero","Butter Cream")
order_2.cake_details()
print("(3)**********************************")
Cakes.give_feedbacks("Chocolate Cake","Very Delicious")
Cakes.give_feedbacks("Chocolate Cake","Yummy")
print("(4)**********************************")
Cakes.show_feedbacks()
print("(5)**********************************")
ch_order1=Cheese_Cakes("Red velvet",700)
ch_order1.cake_details()
print("(6)**********************************")
ch_order1.add_customization()
print("(7)**********************************")
ch_order2=Cheese_Cakes("Blue Berry",900,"No Bake")
ch_order2.cake_details()
print("(8)**********************************")
print(Cakes.order_list)
print("(9)**********************************")
Cheese_Cakes.give_feedbacks("Red velvet Cheese Cake","Average")
Cheese_Cakes.give_feedbacks("Blue Berry Cheese Cake","Liked it")
print("(10)**********************************")
Cakes.show_feedbacks()

```
