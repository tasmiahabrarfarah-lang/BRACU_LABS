```python
#2
inp=input()
l=[]
r=[]
new=""
for i in range(len(inp)):
  if inp[i]==",":
    l+=[int(new)]
    new=""
  elif inp[i] not in "[]":
    new+=inp[i]
l+=[int(new)]
for j in range(len(l)-1,-1,-1):
  r+=[l[j]]
print(r)
```

```python
#7
l=[]
def nested_to_linear(inp):
  for i in inp:
    if type(i)==list:
      nested_to_linear(i)
    else:
      l.append(i)
  return l
nested_to_linear(["start", 10, [4, 2, [11, [9, "mid", 3, [1, 0], 6]], 8], "Done"])
```
