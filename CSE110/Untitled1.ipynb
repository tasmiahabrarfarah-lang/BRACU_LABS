```python
#class evaluation task 1
hour=int(input())
if hour<=168 and hour>=0:
  if hour<=40:
    salary=200*hour
    print(salary)
  else:
    sal=hour-40
    sal=sal*300
    salary=8000+sal
    print(salary)
if hour<0:
  print("Hour cannot be negative")
if hour>168:
  print("Impossible to work more than 168 hours weekly")
```

```python
#class evaluation task 3
distance=int(input())
time=int(input())
distance=distance/1000
time=time/3600
velocity=distance/time
print(velocity,"km/h")
if velocity<60:
  print("Too slow. It needs more changes.")
if velocity<=90 and velocity>=60:
  print("Velocity is okay. The car is ready!")
if velocity>90:
  print("Too fast. Only a few changes should suffice.")
```
