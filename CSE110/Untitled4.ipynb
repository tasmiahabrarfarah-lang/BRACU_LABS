```python
#3.3
def circle(r):
  area=3.1416*r**2
  return area
def sphere(r):
  volume=4/3*3.1416*r**3
  return volume
def fitting(x,y,z):
  if x!=y:
    if z==3:
      volume1=sphere(x/2)
      volume2=sphere(y/2)
      if volume1>volume2:
        left=volume1-volume2
        print(f"Sphere-2 can fit inside Sphere-1 and {left} cubic units would be left.")
      else:
        left=volume2-volume1
        print(f"Sphere-1 can fit inside Sphere-2 and {left} cubic units would be left.")
    else:
      area1=circle(x/2)
      area2=circle(y/2)
      if area1>area2:
        left=area1-area2
        print(f"Circle-2 can fit inside Circle-1 and {left} square units would be left.")
      else:
        left=area2-area1
        print(f"Circle-1 can fit inside Circle-2 and {left} square units would be left.")
  else:
    print("Impossible to fit")
fitting(8,10,2)
fitting(30,14,3)
fitting(5,5,3)
```

```python
#5.3
def recursive_summation(st, n):
  sum=0
  if st!=n:
    sum+=recursive_summation(st+1, n)
  elif st == n:
    return sum
recursive_summation(1,4)
recursive_summation(1,12)
```

```python
#1.3
def is_prime(n):
  prime=False
  count=0
  for i in range(1,n+1):
    if n%i==0:
      count+=1
  if count==2:
    prime=True
  else:
    prime=False
  return prime
def is_perfect(n):
  perfect=False
  sum=0
  for i in range(1,n+1):
    if n%i==0:
      sum+=i
  if sum-n==n:
    perfect=True
  else:
    perfect=False
  return perfect
def special_sum(n):
  sum=0
  for i in range(n+1):
    prm=is_prime(i)
    prfct=is_perfect(i)
    if prm==True or prfct==True:
      sum+=i
  return sum
result = special_sum(8)
print(result)
result = special_sum(30)
print(result)

```

```python
#2.3
def show_dots(n):
  for i in range(n):
    print(".",end="")
def show_palindrome(n):
  for i in range(1,n+1):
    print(i,end="")
  for i in range(n-1,0,-1):
    print(i,end="")
def show_triangle(n):
  for i in range(1,n+1):
    show_dots(n-i)
    show_palindrome(i)
    show_dots(n-i)
    print("\n")
show_triangle(5)
```

```python
#5.0
def sequence_iterative(N):
  sum=0
  for i in range(1,N+1):
    if i%2==0:
      sum-=i
    else:
      sum+=i
  return sum
sequence_iterative(10)
```
