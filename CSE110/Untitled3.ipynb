```python
#class evaluation task 2
inp=input()
vow="AEIOUaeiou"
count_v=0
count_c=0
for i in range(len(inp)):
  if inp[i] in vow:
    count_v+=1
  elif 65<=ord(inp[i])<=90 or 97<=ord(inp[i])<=122:
    count_c+=1
if count_v%3!=0 or count_c%5!=0:
  print("Blimey! No Plunder!!")
else:
  print("Aaarr! Me Plunder!!")
```
